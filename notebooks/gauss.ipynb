```python
import torch

# Ground truth (mean of the Gaussian)
ground_truth = torch.tensor([[5.0, 2.0], [1.0, 3.0]])  # Shape: [batch_size, 2]

# Predicted values
predicted = torch.tensor([[5.2, 2.1], [1.1, 3.1]])  # Shape: [batch_size, 2]

# Create the covariance matrix for independent variables
std_dev = 1.0
covariance_matrix = torch.diag(torch.tensor([std_dev ** 2, std_dev ** 2]))

# Define the multivariate Gaussian for each ground truth
gaussian = torch.distributions.MultivariateNormal(
    loc=ground_truth, covariance_matrix=covariance_matrix
)

# Compute the probabilities of the predicted points
raw_probabilities = torch.exp(gaussian.log_prob(predicted))  # Shape: [batch_size]

# Print results
print("Probabilities:", raw_probabilities)

max_probabilities = torch.exp(gaussian.log_prob(ground_truth))  # Shape: [batch_size]

# Normalize probabilities to make p=1 when predicted equals ground truth
normalized_probabilities = raw_probabilities / max_probabilities

print("Normalized Probabilities:", normalized_probabilities)
```

```python
import random

def biased_random_symmetric(x, std_factor=0.5):
    std = x * std_factor  # Spread relative to X
    return max(-x, min(x, random.gauss(0, std)))  # Clip to [-X, X]

# Example usage
X = 10
value = biased_random_symmetric(X)
print(value)  # More likely close to 0, but still within [-10, 10]
```

```python
import random

def random_symmetric(x):
    return random.uniform(-x, x)


max_perturb_x = 6
value = random_symmetric(max_perturb_x)
print(value)  
```
